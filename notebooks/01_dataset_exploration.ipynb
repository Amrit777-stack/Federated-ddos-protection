```python
# Required libraries
import pandas as pd
import numpy as np

# Load dataset
df = pd.read_csv(
    "../data/raw/CSV-01-12/01-12/Syn.csv",
    nrows=100000
)

# Clean column names
df.columns = df.columns.str.strip()

# Handle missing and infinite values
df["Flow Bytes/s"] = df["Flow Bytes/s"].fillna(0)

df["Flow Packets/s"] = df["Flow Packets/s"].replace(
    [np.inf, -np.inf], np.nan
)

df["Flow Packets/s"] = df["Flow Packets/s"].fillna(0)

# Display basic information
print("Dataset loaded successfully!")
print("Shape:", df.shape)
```

```python
import os

data_path = "../data/raw/CSV-01-12/01-12"

for file in os.listdir(data_path):
    if file.endswith(".csv"):
        temp = pd.read_csv(
            os.path.join(data_path, file),
            usecols=["Label"]
        )
        print(file, ":", temp["Label"].value_counts().to_dict())
```

```python
import os

data_path = "../data/raw/CSV-01-12/01-12"

for file in os.listdir(data_path):
    if file.endswith(".csv"):
        temp = pd.read_csv(os.path.join(data_path, file))
        temp.columns = temp.columns.str.strip()
        
        print("\n", file)
        print(temp["Label"].value_counts().to_dict())
```

```python
import os

data_path = "../data/raw/CSV-01-12/01-12"

for file in os.listdir(data_path):
    if file.endswith(".csv"):
        temp = pd.read_csv(
            os.path.join(data_path, file),
            nrows=1
        )
        label_col = [c for c in temp.columns if c.strip() == "Label"][0]

        counts = {}
        for chunk in pd.read_csv(
            os.path.join(data_path, file),
            usecols=[label_col],
            chunksize=100000
        ):
            chunk.columns = ["Label"]
            counts.update(chunk["Label"].value_counts().to_dict())

        print(file, ":", counts)
```

```python
import os

data_path = "../data/raw/CSV-01-12/01-12"

for file in os.listdir(data_path):
    if file.endswith(".csv"):
        temp = pd.read_csv(
            os.path.join(data_path, file),
            usecols=lambda c: c.strip() == "Label"
        )
        temp.columns = ["Label"]

        print("\n" + file)
        print(temp["Label"].value_counts())
```
